#### Split the data

```python
#select x and y
x = df.drop('test_result', axis=1)
y = df['test_result']
```

```python
from sklearn.model_selection import train_test_split

#split the data into train and test sets
x_train, x_test, y_train, y_test = train_test_split(x,y, train_size=0.7, random_state=1234)
```

#### build the model

```python
from sklearn.linear_model import LogisticRegressionCV

#create an empty model
model= LogisticRegressionCV()

#train the model
model.fit(x_train,y_train)
```

#### model evaluation

```python
y_pred = model.predict(x_test)
y_pred
```

```python
x_test.head()
```

```python
#observed data
y_test.head()
```

```python
#predicted data
y_pred
```

##### model correctly predicted 1st and 2nd , for 3rd row ,observed/given value is 0 but model predicted value is 1, means model faild to predict this,
##### so few correct and in correct, model is not 100 % accurate so we can calculate accuracy, by doing model evaluation

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)
cm
```

```python
tn, fp, fn, tp = cm.ravel()

print(f"true negative = {tn}")
print(f"true positive = {tp}")
print(f"false negative = {fn}")
print(f"false positive = {fp}")
```

#### accuracy

```python
accuracy = (tp+tn) / (tp+tn+fp+fn)
print(f"accuracy = {accuracy *100:.2f}%")
```

#### precision

```python
precision = tp /(tp+fp)
print(f"precision = {precision*100:.2f}%")
```

```python
from sklearn.metrics import precision_score

precision = precision_score(y_test,y_pred)
print(f"precision = {precision * 100:.2f}")
```

#### recall

```python
recall = tp/(tp+fn)
print(f"recall = {recall*100:.2f}%")
```

```python
from sklearn.metrics import recall_score

recall=recall_score(y_test,y_pred)
print(f"recall = {recall*100:.2f}%")
```

#### f1 score

```python
f1 = 2*((precision*recall)/(precision+recall))
print(f"f1 score = {f1*100:.2f}%")
```

```python
from sklearn.metrics import f1_score

f1=f1_score(y_test,y_pred)
print(f"f1 score = {f1*100:.2f}%")
```

#### classification report

```python
from sklearn.metrics import classification_report

print(classification_report(y_test,y_pred))
```

#### result visualization

```python
plt.scatter(x_test['age'][y_pred ==1 ],x_test['physical_score'][y_pred==1], color='red',label="positive")
plt.scatter(x_test['age'][y_pred ==0], x_test['physical_score'][y_pred==0], color='green', label="negative")
plt.xlabel("Age")
plt.ylabel("Physical Score")
plt.title("classification report")
plt.legend()
```
